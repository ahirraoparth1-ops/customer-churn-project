```python
#import the database 
#this is object to connect the sqlite3 to python 
conn = sqlite3.connect('customer_churn.db')
#it will take all name from the database 
sql_query= """ 
    Select name 
    FROM sqlite_master 
    WHERE TYPE= 'table'

""" 

#we will load the data
#here the read_sql is the funtion of panda to read the sql 
#and the paramets given is the query, & connection 
tables= pd.read_sql(sql_query,conn)

#create Table 
#the table inside the conn will store in the tables var
#and then we will convert it into pandas

#create dataframe for each table
for table_name in tables['name']: 
    df=pd.read_sql(f'Select * From {table_name}',conn)
    globals()[f"df_{table_name}"]=df
    print(f"Created dataframe : df_{table_name}" )

conn.close()
```

```python
# SECOND ANALYSIS: which type of plan does people leaving [CHRUN BY PLAN TYPE]

churn_by_plan= (df.groupby('plan_type')['chrun_flag'].mean().mul(100).round(2).reset_index(name='churn_rate_pct'))

print(churn_by_plan)
```

```python
# 4 Churn by subscription Type
df.columns
```

```python
chrun_by_subscri=(df.groupby('subscription_type')['chrun_flag'].mean().mul(100).round(2).reset_index(name='churn_rate_pct'))
```

```python
churn_by_subscription = (
    df.groupby("subscription_type")
      .agg(
          churn_rate_pct=("chrun_flag", lambda s: s.mean() * 100),
          monthly_charges=("monthly_charges", "sum"),
          total_count=('customerid',"count")
      )
      .round(2)
      .reset_index()
)
churn_by_subscription
```

```python
#chrun by state+sum(revenue) & count of users 

churn_by_state=(
    df.groupby("state")
     .agg(
         churn_rate_pct=('chrun_flag', lambda s: s.mean()*100),
         monthly_charges=("monthly_charges", "sum"),
         total_count=('customerid',"count")
     )
    .round(2)
    .reset_index()
)
churn_by_state
```

```python
#6: cal avg tenure 

#cal from 'subscription_start_date' to 'cancellation_date' 
#otherwise use current data if cancellation_date is not there
today= pd.Timestamp.today()
today

df['tenure_days']= np.where(
           df['cancellation_date'].notna(),

    (df['cancellation_date']-df['subscription_start_date']).dt.days, #only happens when con true
    (today - df['subscription_start_date'] ).dt.days

)

df.head()

avg_tenure = df['tenure_days'].mean()
avg_tenure
```

```python
from pathlib import Path
import sqlite3

output_dir = Path("data/processed")
output_dir.mkdir(parents=True, exist_ok=True)

db_path = output_dir / "customer_churn_cleaned.db"

with sqlite3.connect(db_path) as conn:
    df_db_customer.to_sql("customer", conn, if_exists="replace", index=False)
    df_db_subscription.to_sql("subscription", conn, if_exists="replace", index=False)
    df_db_support.to_sql("support", conn, if_exists="replace", index=False)
    df.to_sql("churn_analysis", conn, if_exists="replace", index=False)

df.to_csv(output_dir / "churn_analysis.csv", index=False)

print(f"Saved cleaned database to: {db_path}")
```
